```python
import re 
import requests
```

```python
def read_url(url): 
    pageSource = requests.get(url).text
    
    return pageSource
```

```python
dataSet = list()
URL = "https://godfreysfeed.com/dealersandlocations"
contents = read_url(URL)
```

```python
pattern1 = r'var latLng = new google.maps.LatLng\((?P<lat>.*)\, \s*(?P<lng>.*)\)'
pattern2 = r'infoWindowContent = infoWindowContent\+\s*\"(.*?)\"\;'
```

```python
latlngs = re.findall(pattern1, contents)
```

```python
len(latlngs)
```

```python
dealers = re.findall(pattern2, contents)
```

```python
d = 0 
for dealer in dealers: 
    dealerInfo = re.split(r"<br>", re.sub(r"<br><br>", '', dealer))
    name = re.findall(r'\'>(.*?)</span', dealerInfo[0])[0]
    address = re.findall(r'>(.*)<', dealerInfo[1])[0]
    city = re.findall(r'>(.*),\s*(.*)<', dealerInfo[2])[0][0]
    state = re.findall(r'>(.*),\s*(.*)<', dealerInfo[2])[0][1]
    zip = re.findall(r'>(.*)<', dealerInfo[3])[0]
    lat = latlngs[d][0]
    lng = latlngs[d][1]
    d+=1 
    dataSet.append([name, address, city, state, zip, lat, lng])
```
