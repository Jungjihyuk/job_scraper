```python
import nest_asyncio
import asyncio
from playwright.async_api import async_playwright

nest_asyncio.apply()  # Jupyter에서 asyncio 재사용 허용

async def main():
    async with async_playwright() as play:
        browser = await play.chromium.launch()
        page = await browser.new_page()
        await page.goto("https://www.wanted.co.kr/wd/295402")
        await page.wait_for_selector('main.JobDetail_jobDetail__si4a2 div.JobDetail_relativeWrapper__z7ugu')
        title_elem = await page.query_selector("div.JobDetail_relativeWrapper__z7ugu h1.wds-58fmok")
        title = await title_elem.inner_text() if title_elem else "제목 없음"
        print("Title:", title)
        await browser.close()

await main()
```
