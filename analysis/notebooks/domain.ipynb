## [1] n-gram 빈도 분석

````python
import re
from collections import Counter

# 본문 텍스트 결합 (position_detail + main_tasks + requirements + preferred)
def combine_body(row):
    parts = []
    for col in ['position_detail', 'main_tasks', 'requirements', 'preferred']:
        text = row.get(col)
        if isinstance(text, str) and text.strip():
            parts.append(text)
    return ' '.join(parts)

# 전처리: 불릿, 특수문자 정리 / 영어는 소문자 통일
def clean_text(text: str) -> str:
    text = re.sub(r'[•·▶▸◆■□★☆\-–—]', ' ', text)       # 불릿/대시 제거
    text = re.sub(r'[\(\)\[\]{}]', ' ', text)              # 괄호 제거
    text = re.sub(r'[,\.\!\?;:…\n\r\t]', ' ', text)       # 구두점/개행 → 공백
    text = re.sub(r'\s{2,}', ' ', text).strip()            # 다중 공백 정리
    return text
````

````python
job_posting_data['body'] = job_posting_data.apply(combine_body, axis=1)
job_posting_data['body_clean'] = job_posting_data['body'].apply(clean_text)
````

````python
print(f"총 {len(job_posting_data)}건, 본문 비어있는 건수: {(job_posting_data['body_clean'].str.len() == 0).sum()}")
````

````python
from sklearn.feature_extraction.text import CountVectorizer

# 한국어+영어 혼합 토큰 패턴: 한글 2글자 이상 또는 영문/숫자 조합
# (예: "데이터", "AI", "LLM", "K8s", "CI/CD" 등)
TOKEN_PATTERN = r'[가-힣]{2,}|[A-Za-z][A-Za-z0-9/\.\-]{1,}'

corpus = job_posting_data['body_clean'].tolist()

results = {}
for n, label in [(1, 'unigram'), (2, 'bigram'), (3, 'trigram')]:
    vectorizer = CountVectorizer(
        token_pattern=TOKEN_PATTERN,
        ngram_range=(n, n),
        max_features=5000,
        lowercase=True,
    )
    X = vectorizer.fit_transform(corpus)
    freqs = dict(zip(vectorizer.get_feature_names_out(), X.sum(axis=0).A1.astype(int)))
    results[label] = Counter(freqs).most_common(80)
````

### 문서별 n-gram 빈도 분석
전체 코퍼스 빈도가 아닌, **개별 문서(공고)마다 가장 많이 등장한 키워드**를 뽑고  
그 Top-1 / Top-3 키워드를 전체에서 집계하여 어떤 키워드가 가장 많은 공고의 핵심어로 선정되었는지 확인한다.

````python
# 각 문서별로 CountVectorizer 결과에서 Top-N 빈도 키워드 추출
from collections import Counter

doc_freq_keywords = {}

for n, label in [(1, 'unigram'), (2, 'bigram'), (3, 'trigram')]:
    vectorizer = CountVectorizer(
        token_pattern=TOKEN_PATTERN,
        ngram_range=(n, n),
        max_features=5000,
        lowercase=True,
    )
    X = vectorizer.fit_transform(corpus)
    feature_names = vectorizer.get_feature_names_out()

    top_n = 5
    per_doc = []
    for doc_idx in range(X.shape[0]):
        row = X[doc_idx].toarray().flatten()
        top_indices = row.argsort()[::-1][:top_n]
        keywords = [(feature_names[i], int(row[i])) for i in top_indices if row[i] > 0]
        per_doc.append(keywords)

    doc_freq_keywords[label] = per_doc

# 결과 확인: 첫 5개 문서의 bigram Top-5
for i in range(5):
    pos = job_posting_data.iloc[i]['position']
    kws = doc_freq_keywords['bigram'][i]
    print(f"[{i}] {pos}")
    for rank, (kw, cnt) in enumerate(kws, 1):
        print(f"    {rank}. {kw} (빈도: {cnt})")
    print()
````

````python
# 문서별 Top-1 / Top-3 키워드를 전체에서 집계
for label in ['unigram', 'bigram', 'trigram']:
    per_doc = doc_freq_keywords[label]

    top1 = [kws[0][0] for kws in per_doc if kws]
    top1_counter = Counter(top1).most_common(30)

    top3 = [kw for kws in per_doc for kw, _ in kws[:3]]
    top3_counter = Counter(top3).most_common(30)

    print(f"\n{'='*60}")
    print(f"  [{label}] 문서별 Top-1 빈도 키워드 집계 (상위 30)")
    print(f"{'='*60}")
    for kw, cnt in top1_counter:
        print(f"  {cnt:>4}건  {kw}")

    print(f"\n  [{label}] 문서별 Top-3 빈도 키워드 집계 (상위 30)")
    print(f"  {'-'*56}")
    for kw, cnt in top3_counter:
        print(f"  {cnt:>4}건  {kw}")
````

### TF-IDF 기반 키워드 추출
각 채용공고를 하나의 문서로 보고 TF-IDF를 계산하면,  
모든 공고에 공통으로 등장하는 상용구("경험", "개발", "서비스" 등)는 IDF가 낮아 자연스럽게 걸러지고,  
특정 도메인/기술에서만 등장하는 키워드가 높은 점수를 받는다.

````python
len(corpus)
````

````python
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np

# bigram 기준 TF-IDF (가장 유의미한 n-gram 단위)
vectorizer = TfidfVectorizer(
    token_pattern=TOKEN_PATTERN,
    ngram_range=(1, 3),   # unigram ~ trigram 모두 포함
    max_features=8000,
    lowercase=True,
    min_df=2,
    max_df=0.85,
)

tfidf_matrix = vectorizer.fit_transform(corpus)
feature_names = vectorizer.get_feature_names_out()

print(f"문서 수: {tfidf_matrix.shape[0]}, 피처 수: {tfidf_matrix.shape[1]}")

# 각 문서별 Top 10 키워드 추출
TOP_N = 10
doc_keywords = []

for doc_idx in range(tfidf_matrix.shape[0]):
    row = tfidf_matrix[doc_idx].toarray().flatten()
    top_indices = row.argsort()[::-1][:TOP_N]
    keywords = [(feature_names[i], round(row[i], 4)) for i in top_indices if row[i] > 0]
    doc_keywords.append(keywords)

# DataFrame으로 정리: 각 문서의 position + top keywords
doc_kw_df = pd.DataFrame({
    'posting_id': job_posting_data['posting_id'],
    'position': job_posting_data['position'],
    'top_keywords': [', '.join(f'{kw}({s})' for kw, s in kws) for kws in doc_keywords],
})

doc_kw_df.head(20)
````

````python
# 전체 문서에서 Top-1 키워드만 모아서 빈도 집계
# → "어떤 키워드가 가장 많은 공고에서 핵심 키워드로 선정되었는가"
from collections import Counter

top1_keywords = [kws[0][0] for kws in doc_keywords if kws]
top1_counter = Counter(top1_keywords).most_common(30)

print("각 문서의 Top-1 키워드 빈도 (가장 많은 공고에서 1등으로 뽑힌 키워드)")
print("="*60)
for kw, cnt in top1_counter:
    print(f"  {cnt:>4}건  {kw}")

# Top-3 키워드까지 확장
top3_keywords = [kw for kws in doc_keywords for kw, _ in kws[:3]]
top3_counter = Counter(top3_keywords).most_common(50)

print(f"\n각 문서의 Top-3 키워드 빈도 (상위 50)")
print("="*60)
for kw, cnt in top3_counter:
    print(f"  {cnt:>4}건  {kw}")
````

````python
# 샘플 확인: 특정 position과 그 공고의 Top 키워드를 나란히 보기
# → 키워드가 실제 도메인/기술을 잘 반영하는지 직접 확인

sample_indices = [0, 10, 50, 100, 200, 500, 800, 1000, 1200, 1400]
sample_indices = [i for i in sample_indices if i < len(doc_keywords)]

for idx in sample_indices:
    pos = job_posting_data.iloc[idx]['position']
    kws = doc_keywords[idx]
    print(f"\n[{idx}] {pos}")
    for rank, (kw, score) in enumerate(kws[:7], 1):
        print(f"    {rank}. {kw} ({score})")
````

````python
# 문서별 최빈도 키워드 vs TF-IDF 상위 키워드 비교 DataFrame
# bigram 기준 (가장 의미 있는 n-gram 단위)

comparison_rows = []
for doc_idx in range(len(corpus)):
    freq_kws = doc_freq_keywords['bigram'][doc_idx]
    tfidf_kws = doc_keywords[doc_idx]

    comparison_rows.append({
        'posting_id': job_posting_data.iloc[doc_idx]['posting_id'],
        'position': job_posting_data.iloc[doc_idx]['position'],
        # 최빈도 Top-3
        'freq_1': f"{freq_kws[0][0]} ({freq_kws[0][1]})" if len(freq_kws) > 0 else '',
        'freq_2': f"{freq_kws[1][0]} ({freq_kws[1][1]})" if len(freq_kws) > 1 else '',
        'freq_3': f"{freq_kws[2][0]} ({freq_kws[2][1]})" if len(freq_kws) > 2 else '',
        # TF-IDF Top-3
        'tfidf_1': f"{tfidf_kws[0][0]} ({tfidf_kws[0][1]})" if len(tfidf_kws) > 0 else '',
        'tfidf_2': f"{tfidf_kws[1][0]} ({tfidf_kws[1][1]})" if len(tfidf_kws) > 1 else '',
        'tfidf_3': f"{tfidf_kws[2][0]} ({tfidf_kws[2][1]})" if len(tfidf_kws) > 2 else '',
    })

comparison_df = pd.DataFrame(comparison_rows)
pd.set_option('display.max_colwidth', 30)
pd.set_option('display.width', 200)
````

## [2] 형태소 분석을 통한 키워드 추출

````python
from kiwipiepy import Kiwi

kiwi = Kiwi()

# 명사 태그: NNG(일반명사), NNP(고유명사), SL(외국어), SH(한자)
NOUN_TAGS = {'NNG', 'NNP', 'SL', 'SH'}

def extract_nouns(text: str) -> str:
    """Kiwi로 명사+외국어만 추출하여 공백으로 연결"""
    if not text or not text.strip():
        return ''
    tokens = kiwi.tokenize(text)
    nouns = [t.form for t in tokens if t.tag in NOUN_TAGS and len(t.form) >= 2]
    return ' '.join(nouns)
````

````python
# 형태소(명사) n-gram 빈도 분석: body 합산 vs position_detail
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# 명사 추출 후에는 이미 토큰이 분리되어 있으므로 공백 기준 토큰화
NOUN_TOKEN = r'\S+'

for corpus_label, corpus_data in [('body 합산', corpus_nouns), ('position_detail', corpus_nouns_pd)]:
    print(f"\n{'#'*70}")
    print(f"  [{corpus_label}] 명사 n-gram 빈도 분석")
    print(f"{'#'*70}")

    for n, label in [(1, 'unigram'), (2, 'bigram'), (3, 'trigram')]:
        vec = CountVectorizer(
            token_pattern=NOUN_TOKEN,
            ngram_range=(n, n),
            max_features=5000,
            lowercase=True,
        )
        X = vec.fit_transform(corpus_data)
        freqs = dict(zip(vec.get_feature_names_out(), X.sum(axis=0).A1.astype(int)))
        top40 = Counter(freqs).most_common(40)

        print(f"\n{'='*60}")
        print(f"  [{corpus_label}] Top 40 명사 {label}s")
        print(f"{'='*60}")
        df_top = pd.DataFrame(top40, columns=['keyword', 'count'])
        print(df_top.to_string(index=False))
````

````python
# 형태소(명사) 문서별 빈도 키워드 + TF-IDF 키워드 추출
# body 합산 / position_detail 두 코퍼스 모두 수행

for corpus_label, corpus_data in [('body 합산', corpus_nouns), ('position_detail', corpus_nouns_pd)]:
    print(f"\n{'#'*70}")
    print(f"  [{corpus_label}] 명사 문서별 Top-1/Top-3 빈도 키워드 집계")
    print(f"{'#'*70}")

    for n, label in [(1, 'unigram'), (2, 'bigram')]:
        vec = CountVectorizer(
            token_pattern=NOUN_TOKEN,
            ngram_range=(n, n),
            max_features=5000,
            lowercase=True,
        )
        X = vec.fit_transform(corpus_data)
        fnames = vec.get_feature_names_out()

        per_doc = []
        for doc_idx in range(X.shape[0]):
            row = X[doc_idx].toarray().flatten()
            top_indices = row.argsort()[::-1][:5]
            keywords = [(fnames[i], int(row[i])) for i in top_indices if row[i] > 0]
            per_doc.append(keywords)

        top1 = [kws[0][0] for kws in per_doc if kws]
        top3 = [kw for kws in per_doc for kw, _ in kws[:3]]

        print(f"\n  [{label}] 문서별 Top-1 집계 (상위 30)")
        print(f"  {'='*56}")
        for kw, cnt in Counter(top1).most_common(30):
            print(f"  {cnt:>4}건  {kw}")

        print(f"\n  [{label}] 문서별 Top-3 집계 (상위 30)")
        print(f"  {'-'*56}")
        for kw, cnt in Counter(top3).most_common(30):
            print(f"  {cnt:>4}건  {kw}")
````

````python
# 형태소(명사) TF-IDF: body 합산 / position_detail
doc_keywords_noun = {}

for corpus_label, corpus_data in [('body', corpus_nouns), ('pd', corpus_nouns_pd)]:
    tfidf = TfidfVectorizer(
        token_pattern=NOUN_TOKEN,
        ngram_range=(1, 3),
        max_features=8000,
        lowercase=True,
        min_df=2,
        max_df=0.85,
    )
    X = tfidf.fit_transform(corpus_data)
    fnames = tfidf.get_feature_names_out()
    print(f"[{corpus_label}] 문서 수: {X.shape[0]}, 피처 수: {X.shape[1]}")

    per_doc = []
    for doc_idx in range(X.shape[0]):
        row = X[doc_idx].toarray().flatten()
        top_indices = row.argsort()[::-1][:10]
        keywords = [(fnames[i], round(row[i], 4)) for i in top_indices if row[i] > 0]
        per_doc.append(keywords)

    doc_keywords_noun[corpus_label] = per_doc

    # Top-1 / Top-3 집계
    top1 = [kws[0][0] for kws in per_doc if kws]
    top3 = [kw for kws in per_doc for kw, _ in kws[:3]]

    print(f"\n[{corpus_label}] 명사 TF-IDF 문서별 Top-1 (상위 30)")
    print("="*60)
    for kw, cnt in Counter(top1).most_common(30):
        print(f"  {cnt:>4}건  {kw}")

    print(f"\n[{corpus_label}] 명사 TF-IDF 문서별 Top-3 (상위 50)")
    print("="*60)
    for kw, cnt in Counter(top3).most_common(50):
        print(f"  {cnt:>4}건  {kw}")
    print()
````

## [3] SBERT 기반 Zero-shot 도메인 분류

SBERT로 키워드를 추출하는 대신, **도메인 설명 텍스트**와 **채용공고 텍스트**를 동일 임베딩 공간에 매핑하여  
코사인 유사도로 직접 도메인을 분류한다.

### 장점
- 학습 데이터 없이 도메인 설명만으로 즉시 분류 가능 (zero-shot)
- rule-based처럼 정확한 키워드 일치가 필요 없음 — 의미적으로 유사하면 매칭됨
- 새로운 도메인 추가가 쉬움 (설명 텍스트만 추가하면 됨)

### 한계
- **도메인 목록이 데이터 기반이 아님**: 10개 도메인을 사람이 임의로 정의 → 실제 데이터에 없는 도메인이 포함되거나, 있는 도메인이 누락될 수 있음
- **강제 할당**: 10개 중 하나에 무조건 분류 → 해당 없는 공고도 억지로 분류됨 (유사도 threshold로 완화 가능하나 근본 해결은 아님)
- **도메인 설명 텍스트 의존**: 설명을 어떻게 쓰느냐에 따라 결과가 크게 달라짐 → 결국 사람의 주관에 의존하는 건 rule-based와 동일
- **도메인 간 경계 모호**: AI/딥테크는 거의 모든 공고와 유사도가 높게 나옴 (AI 채용공고 특성상) → 다른 도메인이 묻힘

````python
# 채용공고 인코딩 (position_detail 사용)
# 빈 텍스트는 빈 문자열로 처리
pd_texts = job_posting_data['position_detail'].fillna('').tolist()

# batch 인코딩 (1504건)
doc_embeddings = model.encode(pd_texts, show_progress_bar=True, batch_size=64)
print(f"문서 임베딩 shape: {doc_embeddings.shape}")
````

## [4] SBERT 임베딩 클러스터링 기반 도메인 분류

Zero-shot의 핵심 한계는 **도메인 목록을 사람이 정했다**는 것.  
이를 극복하기 위해, SBERT 임베딩을 클러스터링하여 **데이터가 자연스럽게 형성하는 군집**을 먼저 발견하고,  
각 군집의 특성을 분석하여 도메인 라벨을 부여하는 bottom-up 방식으로 접근한다.

### 흐름
1. 채용공고 임베딩 (이미 계산됨)
2. UMAP으로 차원 축소 → 시각화 가능한 2D 공간으로
3. HDBSCAN으로 클러스터링 → 군집 수를 데이터가 결정 (노이즈 포인트도 허용)
4. 각 클러스터의 대표 키워드 추출 → 클러스터가 어떤 도메인인지 해석
5. Zero-shot 결과와 교차 검증

### 장점
- 도메인 목록을 사전에 정의할 필요 없음 — 데이터가 군집 수와 구조를 결정 (bottom-up)
- HDBSCAN이 노이즈 포인트를 허용하므로, 어디에도 속하지 않는 공고를 억지로 분류하지 않음
- 기존에 예상하지 못했던 도메인 군집을 발견할 가능성이 있음

### 한계
- **데이터 품질 의존**: position_detail이 부실하거나 상투적이면 임베딩 자체가 무의미하여 클러스터링 품질이 저하됨
- **언어/문체 기반 군집화**: SBERT가 "산업 도메인"이 아닌 "언어/문체 유사성"으로 묶는 경향 — 영문 채용공고끼리 별도 클러스터가 형성되는 현상 (예: 클러스터 5에 영어 공고만 집중)
- **도메인 vs 직무 구분 불가**: "제조업에서 AI 개발" vs "의료에서 AI 개발"을 구분하지 못하고, "AI"라는 공통 키워드의 유사성으로 하나로 묶어버림. 우리가 원하는 건 직무가 아닌 **산업 도메인** 기준 분류인데, 임베딩은 이 둘을 구분하는 표현을 학습한 적이 없음
- **대표 키워드 품질 문제**: 클러스터의 TF-IDF 대표 키워드에 전치사(and, to, the), 회사명 등 도메인과 무관한 토큰이 포함됨
- **SBERT 모델 한계**: `ko-sroberta-multitask`는 범용 문장 유사도 모델로, 채용공고의 산업 도메인을 구분하도록 학습된 적이 없음. 모델 성능에 따라 클러스터링 결과가 크게 달라질 수 있음

### 근본적 한계
통계 기반(TF-IDF, n-gram), zero-shot, 클러스터링 모두 **"이 회사가 어느 산업에 속하는가"라는 도메인 지식이 모델/알고리즘 안에 없다**는 공통 한계를 가진다.  
이를 해결하려면 결국 **정답(라벨)이 개입**해야 한다:
- **방향 A**: LLM 라벨링 — Claude/GPT에게 각 공고를 주고 산업 도메인을 직접 판단시킴 (모델 내부의 도메인 지식 활용)
- **방향 B**: 수동 샘플 라벨링 → 분류기 학습 — 100~200건을 사람이 라벨링한 뒤, SBERT 임베딩 + 간단한 분류기(kNN, logistic regression)로 학습

````python
import umap
import hdbscan

# 1. UMAP 차원 축소 (768차원 → 2D 시각화용 + 클러스터링용)
# 클러스터링에는 좀 더 높은 차원(20D)이 유리하고, 시각화에는 2D가 필요
reducer_2d = umap.UMAP(n_components=2, random_state=42, n_neighbors=15, min_dist=0.1)
embedding_2d = reducer_2d.fit_transform(doc_embeddings)

reducer_cluster = umap.UMAP(n_components=20, random_state=42, n_neighbors=15, min_dist=0.0)
embedding_cluster = reducer_cluster.fit_transform(doc_embeddings)

print(f"2D 시각화용: {embedding_2d.shape}")
print(f"클러스터링용: {embedding_cluster.shape}")
````

````python
# 2. HDBSCAN 클러스터링
# min_cluster_size: 최소 군집 크기 (너무 작으면 과도한 군집, 너무 크면 뭉뚱그림)
clusterer = hdbscan.HDBSCAN(
    min_cluster_size=20,
    min_samples=5,
    metric='euclidean',
)
cluster_labels = clusterer.fit_predict(embedding_cluster)

n_clusters = len(set(cluster_labels)) - (1 if -1 in cluster_labels else 0)
n_noise = (cluster_labels == -1).sum()

print(f"발견된 클러스터 수: {n_clusters}")
print(f"노이즈 포인트: {n_noise}건 ({n_noise/len(cluster_labels)*100:.1f}%)")
print(f"\n클러스터별 문서 수:")
for label in sorted(set(cluster_labels)):
    count = (cluster_labels == label).sum()
    name = "노이즈" if label == -1 else f"클러스터 {label}"
    print(f"  {name}: {count}건")
````

````python
# 3. 2D 시각화: 클러스터별 색상
import matplotlib.pyplot as plt
from matplotlib import rc

rc('font', family='AppleGothic')
plt.rcParams['axes.unicode_minus'] = False

fig, ax = plt.subplots(figsize=(14, 10))

# 노이즈 포인트는 회색으로
noise_mask = cluster_labels == -1
ax.scatter(
    embedding_2d[noise_mask, 0], embedding_2d[noise_mask, 1],
    c='lightgray', s=5, alpha=0.3, label='노이즈'
)

# 각 클러스터 색상 구분
cmap = plt.cm.get_cmap('tab20', n_clusters)
for label in sorted(set(cluster_labels)):
    if label == -1:
        continue
    mask = cluster_labels == label
    count = mask.sum()
    ax.scatter(
        embedding_2d[mask, 0], embedding_2d[mask, 1],
        c=[cmap(label)], s=10, alpha=0.6, label=f'클러스터 {label} ({count}건)'
    )

ax.set_title('SBERT 임베딩 UMAP 2D + HDBSCAN 클러스터링', fontsize=14, fontweight='bold')
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=8)
plt.tight_layout()
plt.show()
````

````python
# 4. 각 클러스터의 대표 키워드 추출 (TF-IDF 기반)
# → 클러스터가 어떤 도메인인지 해석하는 근거

from sklearn.feature_extraction.text import TfidfVectorizer

job_posting_data['cluster'] = cluster_labels

# 클러스터별 문서를 합쳐서 하나의 "대표 문서"로 만든 뒤, 클러스터 간 TF-IDF 비교
# 이렇게 하면 특정 클러스터에서만 자주 등장하는 키워드가 높은 점수를 받음
cluster_docs = []
cluster_ids = []
for label in sorted(set(cluster_labels)):
    if label == -1:
        continue
    mask = cluster_labels == label
    texts = job_posting_data.loc[mask, 'position_detail'].fillna('').tolist()
    cluster_docs.append(' '.join(texts))
    cluster_ids.append(label)

tfidf_cluster = TfidfVectorizer(
    token_pattern=TOKEN_PATTERN,
    ngram_range=(1, 2),
    max_features=5000,
    lowercase=True,
    min_df=1,
    max_df=0.8,
)
X_cluster = tfidf_cluster.fit_transform(cluster_docs)
fnames = tfidf_cluster.get_feature_names_out()

print("각 클러스터의 대표 키워드 Top 15 (position_detail 기준)")
print("="*70)
for i, label in enumerate(cluster_ids):
    count = (cluster_labels == label).sum()
    row = X_cluster[i].toarray().flatten()
    top_indices = row.argsort()[::-1][:15]
    keywords = [fnames[idx] for idx in top_indices]

    print(f"\n클러스터 {label} ({count}건):")
    print(f"  {', '.join(keywords)}")
````

````python
# 5. 각 클러스터의 대표 position(직무명) 확인
# → 키워드만으로 부족할 때, 실제 어떤 공고들이 모였는지 직접 확인

for label in sorted(set(cluster_labels)):
    if label == -1:
        continue
    mask = cluster_labels == label
    positions = job_posting_data.loc[mask, 'position'].tolist()
    count = len(positions)

    print(f"\n{'='*60}")
    print(f"  클러스터 {label} ({count}건) — 대표 직무명 (최대 10개)")
    print(f"{'='*60}")
    # 랜덤 대신 앞에서 10개
    for pos in positions[:10]:
        print(f"  - {pos}")
````

````python
# 6. 클러스터 vs Zero-shot 도메인 교차 분석
# → 클러스터가 Zero-shot 도메인과 얼마나 겹치는지 확인

print("클러스터별 Zero-shot 도메인 분포")
print("="*70)

for label in sorted(set(cluster_labels)):
    if label == -1:
        continue
    mask = cluster_labels == label
    domain_dist = job_posting_data.loc[mask, 'sbert_domain'].value_counts()
    count = mask.sum()

    print(f"\n클러스터 {label} ({count}건):")
    for domain, cnt in domain_dist.items():
        pct = cnt / count * 100
        bar = '█' * int(pct / 2)
        print(f"  {domain:<15} {cnt:>4}건 ({pct:5.1f}%) {bar}")
````

**프롬프트 예시:**

```
labeling_data/ 폴더에 labeling_chunk_00.csv ~ labeling_chunk_10.csv 파일이 있어.
각 CSV에는 posting_id, position, position_detail 컬럼이 있고, 채용공고 데이터야.

각 청크 파일을 병렬 Agent로 나눠서 다음 작업을 해줘:

1. CSV를 읽고 각 행의 position + position_detail을 보고 이 채용공고를 올린 
   **회사의 산업 도메인**을 자유롭게 판단해줘
2. 직무(AI엔지니어, 백엔드 등)가 아니라 **회사가 속한 산업**(의료, 금융, 제조, 게임 등)을 분류해줘

4. 도메인 판단이 어려운 경우 "미분류"로 표시해줘
```

---

- "병렬 Agent로 나눠서"라고 명시해야 Claude Code가 Agent tool을 병렬로 실행
- 도메인 목록을 고정하지 않고 자유 분류 → 전체 결과 취합 후 유사 도메인 통합
